**(c)** Which pandas data type did each column receive? Does `household_id` deserve to be treated as a number in calculations?

**Answer:**

| Column | pandas dtype | Meaning |
|---|---|---|
| `household_id` | `int64` | whole numbers (integers) |
| `household_size` | `int64` | whole numbers (count of people) |
| `monthly_kwh` | `float64` | decimal numbers (continuous measurement) |
| `housing_type` | `str` (shown as `object` in older pandas versions) | text / categorical |
| `income_bracket` | `str` (shown as `object` in older pandas versions) | text / categorical (ordinal) |

**Does `household_id` deserve to be treated as a number? No.** It is only a label (identifier) that names each household. Its numeric value has no quantitative meaning: the mean of the IDs, the sum of the IDs, or saying that household 106 is "larger" than household 101 tells us nothing. Pandas gave it `int64` only because the values look like integers. It should be treated as an identifier (a string/categorical variable), e.g. `task_1_df["household_id"] = task_1_df["household_id"].astype(str)`, so that it is excluded from numeric calculations and statistics.
